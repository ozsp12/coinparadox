# The coin paradox

There is an interesting mathematics problem that became popular in 1975. It was proposed by **Martin Gardner**, a famous mathematical popularizer who created puzzles and challenges [1]. Yes—there are people who solve mathematical problems for fun! Originally, the question was:

> **“How many complete revolutions does a coin make when it goes once around the circumference of another coin of the same size?”**

![texte](https://upload.wikimedia.org/wikipedia/commons/8/80/Coin_reverse_rotation_paradox.gif)

This question became known as the **Coin Rotation Paradox** [2]. It was later reformulated and gained notoriety for being the only question that no one ever answered correctly on the SAT, the American equivalent of the college entrance exam. The problem was that the SAT organizers themselves made a mistake in the solution and did not include the correct answer among the choices [3,4]. The statement was:

> **“How many revolutions does a circle of radius \( r \) make while traveling around the circumference of a circle of radius \( R = 3r \)?”**

The most commonly given answer is **3**. However, the correct answer is **4**, which is rather counterintuitive. Below is a visual demonstration taken from the *Coin Rotation Paradox* entry on Wikipedia [2]. Below is a visual demonstration of how the answer is **4 complete revolutions**.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

# Define the animation update function called at each frame
def update(frame):
    # Extract the current angular position from the theta array
    t = theta[frame]
    # Compute the x-coordinate of the small circle center
    x = (R + r) * np.cos(t)
    # Compute the y-coordinate of the small circle center
    y = (R + r) * np.sin(t)
    # Compute the rotation angle of the small circle (counterclockwise)
    phi = (R + r) * t / r  # counterclockwise

    # Update the center position of the small circle
    small_circle.center = (x, y)
    # Update the position of the red marker on the circumference
    marker.set_data(x + r * np.cos(phi - np.pi), y + r * np.sin(phi - np.pi))

    # Compute the number of completed full rotations
    n_rot = int(phi // (2 * np.pi))
    # Update the text displaying the number of full rotations
    counter_text.set_text(f"Number of complete turns: {n_rot}")

    # Return updated artists for blitting
    return small_circle, marker, counter_text

# Define the radius of the small circle
r = 1
# Define the radius of the large circle as three times the small radius
R = 3 * r
# Define the total number of animation frames
n_frames = 360
# Create an array of angles uniformly spaced from 0 to 2π
theta = np.linspace(0, 2*np.pi, n_frames)
# Create a square figure and axis for plotting
fig, ax = plt.subplots(figsize=(6, 6))
# Enforce equal scaling on both axes to preserve circular shapes
ax.set_aspect('equal')
# Set horizontal axis limits
ax.set_xlim(-5*r, 5*r)
# Set vertical axis limits
ax.set_ylim(-5*r, 5*r)
# Remove axis ticks and frame for a clean visualization
ax.axis('off')
# Create the fixed large circle centered at the origin
big_circle = plt.Circle((0, 0), R, color='black', fill=False, linewidth=2)
# Add the large circle to the plot
ax.add_patch(big_circle)
# Create a static blue marker at the initial contact point
initial_contact = plt.Circle((R, 0), 0.07, color='blue')
# Add the initial contact marker to the plot
ax.add_patch(initial_contact)
# Create the moving small circle (initially centered at the origin)
small_circle = plt.Circle((0, 0), r, color='black', fill=False, linewidth=2)
# Create a red point marker to track rotation on the small circle
marker, = ax.plot([], [], 'ro', markersize=5)
# Add the small circle to the plot
ax.add_patch(small_circle)
# Create a text object to display the rotation counter
counter_text = ax.text(-4.5*r, 4.5*r, '', fontsize=14, color='darkred', weight='bold')

# Create the animation object using the update function
ani = animation.FuncAnimation(fig, update, frames=n_frames, blit=True, interval=40)

# Close the figure to prevent duplicate static display
plt.close(fig)
# Render the animation as HTML for display in Jupyter
HTML(ani.to_jshtml())
# Optionally save the animation to an MP4 file
#ani.save('circle_on_circle.mp4')
```
